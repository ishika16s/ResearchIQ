```python

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"

model = model.to(device)

print("Model loaded successfully!")
print("Device:", device)
```

```python

# Select one research paper

article = df["article_clean"][0]

# Tokenize the article
inputs = tokenizer(
    article,
    max_length=1024,
    truncation=True,
    return_tensors="pt"
)

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}

# Generate summary
with torch.no_grad():
    summary_ids = model.generate(
        inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_length=150,
        min_length=40,
        num_beams=4,
        early_stopping=True
    )

# Convert generated tokens into text
generated_summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print("GENERATED RESEARCH SUMMARY:\n")
print(generated_summary)
```

```python

from datasets import DatasetDict

# Divide the existing 1000 papers
split_dataset = dataset.train_test_split(
    test_size=0.1,
    seed=42
)

train_data = split_dataset["train"]
validation_data = split_dataset["test"]

print("Training examples:", len(train_data))
print("Validation examples:", len(validation_data))
```

```python

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM
)

model_name = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name
)

print("BART model ready for fine-tuning!")
```

```python

max_input_length = 512
max_target_length = 128

def preprocess_function(examples):

    inputs = tokenizer(
        examples["article"],
        max_length=max_input_length,
        truncation=True
    )

    targets = tokenizer(
        text_target=examples["abstract"],
        max_length=max_target_length,
        truncation=True
    )

    inputs["labels"] = targets["input_ids"]

    return inputs


tokenized_train = train_data.map(
    preprocess_function,
    batched=True,
    remove_columns=train_data.column_names
)

tokenized_validation = validation_data.map(
    preprocess_function,
    batched=True,
    remove_columns=validation_data.column_names
)

print("Training tokenization completed!")
print(tokenized_train)
```

```python

from transformers import DataCollatorForSeq2Seq

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model
)

print("Data collator ready!")
```

```python

from transformers import Seq2SeqTrainingArguments

training_args = Seq2SeqTrainingArguments(

    output_dir="./bart_research_model",

    learning_rate=2e-5,

    per_device_train_batch_size=1,

    per_device_eval_batch_size=1,

    gradient_accumulation_steps=4,

    num_train_epochs=1,

    eval_strategy="epoch",

    save_strategy="epoch",

    logging_steps=25,

    save_total_limit=2,

    predict_with_generate=True,

    generation_max_length=128,

    fp16=torch.cuda.is_available(),

    report_to="none"
)

print("Training configuration ready!")
```

```python

from transformers import Seq2SeqTrainer

trainer = Seq2SeqTrainer(

    model=model,

    args=training_args,

    train_dataset=tokenized_train,

    eval_dataset=tokenized_validation,

    data_collator=data_collator
)

print("Trainer initialized successfully!")
```

```python
def split_into_chunks(text, chunk_size=150, overlap=30):

    words = text.split()
    chunks = []

    start = 0

    while start < len(words):

        end = start + chunk_size

        chunk = " ".join(words[start:end])

        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
sample_text = rag_documents[0]

sample_chunks = split_into_chunks(sample_text)

print("Total chunks:", len(sample_chunks))
print("\nFirst chunk:")
print(sample_chunks[0])
```

```python
all_chunks = []

for paper_id, document in enumerate(rag_documents):

    chunks = split_into_chunks(document)

    for chunk_id, chunk in enumerate(chunks):

        all_chunks.append({
            "paper_id": paper_id,
            "chunk_id": chunk_id,
            "text": chunk
        })

print("Total chunks created:", len(all_chunks))
print("\nFirst chunk:")
print(all_chunks[0])
```

```python
chunk_texts = [
    chunk["text"] for chunk in all_chunks
]

chunk_embeddings = embedding_model.encode(
    chunk_texts,
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Chunk embedding shape:", chunk_embeddings.shape)
```

```python
import faiss
import numpy as np

dimension = chunk_embeddings.shape[1]

rag_index = faiss.IndexFlatIP(dimension)

rag_index.add(
    chunk_embeddings.astype("float32")
)

print("RAG FAISS index created!")
print("Total chunks indexed:", rag_index.ntotal)
```

```python
def retrieve_context(question, top_k=3):

    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores, indices = rag_index.search(
        question_embedding.astype("float32"),
        top_k
    )

    retrieved_chunks = []

    for idx, score in zip(indices[0], scores[0]):

        if idx == -1:
            continue

        retrieved_chunks.append({
            "text": all_chunks[idx]["text"],
            "paper_id": all_chunks[idx]["paper_id"],
            "chunk_id": all_chunks[idx]["chunk_id"],
            "similarity": float(score)
        })

    return retrieved_chunks
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

generator_name = "google/flan-t5-base"

generator_tokenizer = AutoTokenizer.from_pretrained(
    generator_name
)

generator_model = AutoModelForSeq2SeqLM.from_pretrained(
    generator_name
)

device = "cuda" if torch.cuda.is_available() else "cpu"

generator_model = generator_model.to(device)

print("Generator loaded on:", device)
```

```python
def generate_rag_answer(question, retrieved_chunks):

    context = "\n\n".join(
        item["text"] for item in retrieved_chunks
    )

    prompt = f"""
    Answer the question using only the given context.

    If the context does not contain the answer,
    say that the information is not available.

    Context:
    {context}

    Question:
    {question}

    Answer:
    """

    inputs = generator_tokenizer(
        prompt,
        max_length=512,
        truncation=True,
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():

        outputs = generator_model.generate(
            **inputs,
            max_new_tokens=150,
            num_beams=4
        )

    answer = generator_tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer
```

```python
def ask_research_assistant(question):

    retrieved = retrieve_context(question, top_k=3)

    if not retrieved:
        return "No relevant research information found."

    answer = generate_rag_answer(
        question,
        retrieved
    )

    print("\nQUESTION:")
    print(question)

    print("\nGENERATED ANSWER:")
    print(answer)

    print("\nSOURCES:")

    for item in retrieved:

        print(
            "Paper ID:", item["paper_id"],
            "| Chunk ID:", item["chunk_id"],
            "| Similarity:", round(item["similarity"], 4)
        )

    return answer
```

```python
def ask_pdf_question(question, top_k=3):

    # Step 1: Retrieve relevant PDF chunks
    retrieved = retrieve_pdf_context(question, top_k)

    if not retrieved:
        print("No relevant information found.")
        return

    # Step 2: Combine retrieved text
    context = "\n\n".join(
        item["text"] for item in retrieved
    )

    # Step 3: Prepare prompt
    prompt = f"""
    Answer the question using only the context below.

    If the answer is not available in the context,
    say that the information is not available.

    Context:
    {context}

    Question:
    {question}

    Answer:
    """

    # Step 4: Tokenize
    inputs = generator_tokenizer(
        prompt,
        max_length=512,
        truncation=True,
        return_tensors="pt"
    ).to(device)

    # Step 5: Generate answer
    with torch.no_grad():

        outputs = generator_model.generate(
            **inputs,
            max_new_tokens=150,
            num_beams=4
        )

    answer = generator_tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    # Step 6: Display answer
    print("\nQUESTION:")
    print(question)

    print("\nGENERATED ANSWER:")
    print(answer)

    # Step 7: Display sources
    print("\nSOURCE PAGES:")

    for item in retrieved:
        print(
            "Page:", item["page_number"],
            "| Similarity:",
            round(item["similarity"], 4)
        )

    return answer
```

```python
def ask_multiple_pdfs(question, top_k=3):

    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores, indices = multi_pdf_index.search(
        query_embedding.astype("float32"),
        min(top_k, multi_pdf_index.ntotal)
    )

    retrieved = []

    for idx, score in zip(indices[0], scores[0]):

        if idx == -1:
            continue

        retrieved.append(all_pdf_chunks[idx])

    if not retrieved:
        print("No relevant information found.")
        return

    context = "\n\n".join(
        item["text"] for item in retrieved
    )

    prompt = f"""
    Answer using only the following research paper context.
    If information is missing, say so.

    Context:
    {context}

    Question:
    {question}

    Answer:
    """

    inputs = generator_tokenizer(
        prompt,
        max_length=512,
        truncation=True,
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():

        outputs = generator_model.generate(
            **inputs,
            max_new_tokens=150,
            num_beams=4
        )

    answer = generator_tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    print("\nANSWER:")
    print(answer)

    print("\nSOURCES:")

    for item in retrieved:
        print(
            item["filename"],
            "| Page:", item["page_number"]
        )

    return answer
```

```python
def precision_at_k(relevant_items, retrieved_items, k=5):

    retrieved = retrieved_items[:k]

    relevant_count = sum(
        item in relevant_items for item in retrieved
    )

    return relevant_count / k
```

```python
def recall_at_k(relevant_items, retrieved_items, k=5):

    retrieved = retrieved_items[:k]

    relevant_count = sum(
        item in relevant_items for item in retrieved
    )

    if len(relevant_items) == 0:
        return 0

    return relevant_count / len(relevant_items)
```

```python
def summarize_research_paper(filename):

    text = assistant.papers[filename]

    inputs = tokenizer(
        text[:3000],
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_length=150,
        min_length=30,
        num_beams=4
    )

    summary = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return summary
```

```python
def get_paper_information(filename):

    text = assistant.papers[filename]

    questions = {
        "Objective": "What is the main objective?",
        "Methodology": "What methodology is used?",
        "Dataset": "What datasets are used?",
        "Results": "What are the main results?",
        "Limitations": "What are the limitations?"
    }

    information = {}

    for category, question in questions.items():

        information[category] = extract_paper_detail(
            text,
            question
        )

    return information
```

```python
def search_papers(query, top_k=5):

    query_embedding = embedding_model.encode(
        [query],
        normalize_embeddings=True
    ).astype("float32")

    scores, indices = multi_pdf_index.search(
        query_embedding,
        min(top_k, multi_pdf_index.ntotal)
    )

    results = []

    for score, index in zip(scores[0], indices[0]):

        if index == -1:
            continue

        chunk = all_pdf_chunks[index]

        results.append({
            "paper": chunk["filename"],
            "page": chunk["page_number"],
            "similarity": float(score),
            "text": chunk["text"]
        })

    return results
```

```python
def classify_paper(filename):

    text = assistant.papers[filename][:5000]

    result = topic_classifier(
        text,
        candidate_labels=topic_labels,
        multi_label=True,
        truncation=True
    )

    return [
        {
            "topic": label,
            "score": round(score, 3)
        }
        for label, score in zip(
            result["labels"][:5],
            result["scores"][:5]
        )
    ]
```

```python
def run_research_assistant():

    print("\n===== AI RESEARCH PAPER INTELLIGENCE SYSTEM =====")

    print("\nAvailable Papers:")

    filenames = list(assistant.papers.keys())

    for i, filename in enumerate(filenames):
        print(i, filename)

    print("\nSelect a feature:")
    print("1. Summarize Paper")
    print("2. Extract Paper Information")
    print("3. Search Research Papers")
    print("4. Classify Paper")
    print("5. Ask a Research Question")

    choice = input("\nEnter choice (1-5): ")

    if choice in ["1", "2", "4"]:

        paper_number = int(input("Enter paper number: "))

        if paper_number < 0 or paper_number >= len(filenames):
            print("Invalid paper number.")
            return

        filename = filenames[paper_number]

    if choice == "1":

        print(summarize_research_paper(filename))

    elif choice == "2":

        information = get_paper_information(filename)

        for key, value in information.items():
            print(f"\n{key}: {value}")

    elif choice == "3":

        query = input("Enter search query: ")

        results = search_papers(query)

        for result in results:
            print("\nPaper:", result["paper"])
            print("Page:", result["page"])
            print("Similarity:", round(result["similarity"], 3))
            print("Passage:", result["text"][:300])

    elif choice == "4":

        print(classify_paper(filename))

    elif choice == "5":

        question = input("Ask your research question: ")

        result = research_question_answering(question)

        print("\nAnswer:", result["answer"])

        print("\nSources:")

        for source in result["sources"]:
            print(source["filename"], "| Page:", source["page_number"])

    else:

        print("Invalid choice.")
```

```python

%%writefile app.py

import streamlit as st
from datetime import datetime

# ---------------- PAGE CONFIG ----------------

st.set_page_config(
    page_title="ResearchIQ | AI Research Workspace",
    page_icon="🔬",
    layout="wide",
    initial_sidebar_state="expanded"
)

# ---------------- SESSION STATE ----------------

if "uploaded_papers" not in st.session_state:
    st.session_state.uploaded_papers = []

if "chat_history" not in st.session_state:
    st.session_state.chat_history = []

# ---------------- CUSTOM DESIGN ----------------

st.markdown("""
<style>
@import url('https://fonts.googleapis.com/css2?family=DM+Sans:wght@400;500;600;700&family=Manrope:wght@400;500;600;700;800&display=swap');

html, body, [class*="css"] {
    font-family: 'DM Sans', sans-serif;
}

.stApp {
    background: #F7F8FC;
}

[data-testid="stSidebar"] {
    background: #111827;
    border-right: 1px solid #263247;
}

[data-testid="stSidebar"] * {
    color: #E5E7EB;
}

.brand {
    font-family: 'Manrope', sans-serif;
    font-size: 27px;
    font-weight: 800;
    color: white;
    letter-spacing: -1px;
    padding: 12px 0 2px 0;
}

.brand span {
    color: #A78BFA;
}

.sidebar-caption {
    color: #9CA3AF;
    font-size: 12px;
    margin-bottom: 28px;
}

.main-title {
    font-family: 'Manrope', sans-serif;
    font-size: 32px;
    font-weight: 800;
    color: #172033;
    letter-spacing: -1px;
    margin-bottom: 3px;
}

.subtitle {
    color: #778196;
    font-size: 14px;
    margin-bottom: 24px;
}

.hero {
    background: linear-gradient(115deg, #202B50 0%, #43358B 58%, #7054C8 100%);
    border-radius: 18px;
    padding: 30px;
    color: white;
    margin: 12px 0 24px 0;
}

.hero h2 {
    color: white;
    font-family: 'Manrope', sans-serif;
    font-size: 27px;
    margin-bottom: 8px;
}

.hero p {
    color: #E1DDFB;
    font-size: 14px;
    max-width: 600px;
}

.section-title {
    color: #202A3D;
    font-family: 'Manrope', sans-serif;
    font-weight: 750;
    font-size: 19px;
    margin: 22px 0 12px 0;
}

.metric-card {
    background: white;
    border: 1px solid #E8EAF2;
    border-radius: 15px;
    padding: 19px;
    min-height: 112px;
}

.metric-label {
    color: #7B8497;
    font-size: 12px;
    font-weight: 600;
}

.metric-value {
    color: #202A3D;
    font-family: 'Manrope', sans-serif;
    font-size: 27px;
    font-weight: 800;
    margin-top: 8px;
}

.feature-card {
    background: white;
    border: 1px solid #E8EAF2;
    border-radius: 15px;
    padding: 20px;
    min-height: 155px;
}

.feature-icon {
    font-size: 23px;
    margin-bottom: 9px;
}

.feature-card h4 {
    color: #202A3D;
    font-size: 15px;
    margin: 0 0 7px 0;
}

.feature-card p {
    color: #7B8497;
    font-size: 12px;
    line-height: 1.6;
}

.info-box {
    background: #F0EDFF;
    border: 1px solid #E2DCFF;
    border-radius: 12px;
    padding: 15px;
    color: #51418C;
    font-size: 13px;
}

div.stButton > button {
    border-radius: 10px;
    border: 1px solid #DDD6FE;
    background: #6D5BD0;
    color: white;
    font-weight: 600;
    min-height: 42px;
}

div.stButton > button:hover {
    background: #5846B6;
    border-color: #5846B6;
    color: white;
}

.stTextInput input, .stTextArea textarea {
    border-radius: 10px;
}

div[data-testid="stFileUploader"] {
    background: white;
    border: 1px dashed #B9B1E8;
    border-radius: 14px;
    padding: 12px;
}

hr {
    border-color: #E8EAF2;
}

.footer {
    color: #929BAD;
    font-size: 11px;
    text-align: center;
    padding: 25px 0 8px 0;
}
</style>
""", unsafe_allow_html=True)

# ---------------- SIDEBAR ----------------

with st.sidebar:

    st.markdown(
        '<div class="brand">Research<span>IQ</span></div>',
        unsafe_allow_html=True
    )

    st.markdown(
        '<div class="sidebar-caption">AI-POWERED RESEARCH WORKSPACE</div>',
        unsafe_allow_html=True
    )

    st.markdown("---")

    page = st.radio(
        "WORKSPACE",
        [
            "Overview",
            "Paper Library",
            "Summarization",
            "Research Copilot",
            "Semantic Search",
            "Paper Comparison",
            "Research Insights"
        ],
        label_visibility="visible"
    )

    st.markdown("---")

    st.markdown("**YOUR WORKSPACE**")

    st.caption(
        f"📚 {len(st.session_state.uploaded_papers)} uploaded file(s)"
    )

    st.markdown("---")
    st.caption("ResearchIQ · Project Prototype")
    st.caption("Built with Python & Transformers")

# ---------------- REUSABLE HEADER ----------------

def page_header(title, description):
    st.markdown(
        f'<div class="main-title">{title}</div>'
        f'<div class="subtitle">{description}</div>',
        unsafe_allow_html=True
    )

# ---------------- OVERVIEW ----------------

if page == "Overview":

    page_header(
        "Research workspace",
        "Your central space to explore, understand and compare research."
    )

    st.markdown("""
    <div class="hero">
        <h2>Turn research papers into insights.</h2>
        <p>
        Upload academic papers, discover important findings,
        ask questions and explore research with AI-powered tools.
        Everything you need for a smarter research workflow.
        </p>
    </div>
    """, unsafe_allow_html=True)

    col1, col2, col3 = st.columns(3)

    metrics = [
        ("PAPERS UPLOADED", len(st.session_state.uploaded_papers)),
        ("AI FEATURES", "6"),
        ("WORKSPACE", "Active")
    ]

    for col, (label, value) in zip([col1, col2, col3], metrics):
        with col:
            st.markdown(f"""
            <div class="metric-card">
                <div class="metric-label">{label}</div>
                <div class="metric-value">{value}</div>
            </div>
            """, unsafe_allow_html=True)

    st.markdown(
        '<div class="section-title">Explore your research tools</div>',
        unsafe_allow_html=True
    )

    features = [
        ("📄", "Paper Summarization",
         "Understand lengthy academic papers through concise summaries."),
        ("💬", "Research Copilot",
         "Ask questions and explore answers grounded in your papers."),
        ("🔎", "Semantic Search",
         "Find relevant research passages using natural language."),
        ("⚖️", "Paper Comparison",
         "Compare objectives, methodologies and findings."),
        ("🧠", "Topic Classification",
         "Discover the research topics covered by your papers."),
        ("📊", "Research Insights",
         "Explore extracted information and research patterns.")
    ]

    for start in range(0, len(features), 3):
        cols = st.columns(3)

        for col, item in zip(cols, features[start:start+3]):
            icon, title, description = item

            with col:
                st.markdown(f"""
                <div class="feature-card">
                    <div class="feature-icon">{icon}</div>
                    <h4>{title}</h4>
                    <p>{description}</p>
                </div>
                """, unsafe_allow_html=True)

    st.markdown(
        '<div class="section-title">Get started</div>',
        unsafe_allow_html=True
    )

    st.markdown("""
    <div class="info-box">
        <b>New to ResearchIQ?</b><br>
        Start by uploading your research papers in the Paper Library.
        Then explore summarization, search and research chat.
    </div>
    """, unsafe_allow_html=True)

# ---------------- PAPER LIBRARY ----------------

elif page == "Paper Library":

    page_header(
        "Paper Library",
        "Upload and organize the research papers you want to explore."
    )

    uploaded_files = st.file_uploader(
        "Upload research papers",
        type=["pdf"],
        accept_multiple_files=True,
        help="Choose one or more PDF research papers."
    )

    if uploaded_files:
        existing_names = {
            file.name for file in st.session_state.uploaded_papers
        }

        for file in uploaded_files:
            if file.name not in existing_names:
                st.session_state.uploaded_papers.append(file)

        st.success("Uploaded files added to your session.")

    st.markdown(
        '<div class="section-title">Your documents</div>',
        unsafe_allow_html=True
    )

    if st.session_state.uploaded_papers:

        for i, file in enumerate(st.session_state.uploaded_papers):
            with st.container(border=True):
                col1, col2 = st.columns([5, 1])

                with col1:
                    st.markdown(f"**📄 {file.name}**")
                    st.caption(
                        f"PDF document · {file.size / 1024:.1f} KB"
                    )

                with col2:
                    if st.button("Remove", key=f"remove_{i}"):
                        st.session_state.uploaded_papers.pop(i)
                        st.rerun()

    else:
        st.info("No papers uploaded yet. Upload your first PDF above.")

# ---------------- SUMMARIZATION ----------------

elif page == "Summarization":

    page_header(
        "Paper Summarization",
        "Turn lengthy academic papers into easier-to-understand summaries."
    )

    st.markdown(
        '<div class="info-box">Select a paper and generate a concise summary. '
        'The summarization model will be connected in the backend integration stage.</div>',
        unsafe_allow_html=True
    )

    if st.session_state.uploaded_papers:

        names = [
            file.name for file in st.session_state.uploaded_papers
        ]

        selected = st.selectbox("Select a research paper", names)

        if st.button("Generate Summary"):
            st.info(
                "Frontend ready. Model connection is the next integration task."
            )

    else:
        st.warning("Upload a PDF in Paper Library to get started.")

# ---------------- RESEARCH COPILOT ----------------

elif page == "Research Copilot":

    page_header(
        "Research Copilot",
        "Ask questions and explore information from your research papers."
    )

    st.markdown("""
    <div class="info-box">
        Ask about a paper's methodology, findings, architecture,
        datasets or conclusions. Answers will include source references
        once the RAG backend is connected.
    </div>
    """, unsafe_allow_html=True)

    for message in st.session_state.chat_history:
        with st.chat_message(message["role"]):
            st.write(message["content"])

    question = st.chat_input("Ask something about your research...")

    if question:
        st.session_state.chat_history.append({
            "role": "user",
            "content": question
        })

        with st.chat_message("user"):
            st.write(question)

        with st.chat_message("assistant"):
            st.info(
                "The research-answering backend will be connected next."
            )

    if st.button("Clear conversation"):
        st.session_state.chat_history = []
        st.rerun()

# ---------------- SEMANTIC SEARCH ----------------

elif page == "Semantic Search":

    page_header(
        "Semantic Search",
        "Search research by meaning instead of relying only on exact keywords."
    )

    query = st.text_input(
        "What would you like to find?",
        placeholder="Example: How does self-attention work?"
    )

    top_k = st.slider(
        "Number of results",
        min_value=1,
        max_value=10,
        value=5
    )

    if st.button("Search Research"):
        if query.strip():
            st.info(
                "Search interface ready. FAISS backend connection is pending."
            )
        else:
            st.warning("Please enter a search query.")

# ---------------- PAPER COMPARISON ----------------

elif page == "Paper Comparison":

    page_header(
        "Paper Comparison",
        "Understand how two research papers differ and where they overlap."
    )

    if len(st.session_state.uploaded_papers) >= 2:

        names = [
            file.name for file in st.session_state.uploaded_papers
        ]

        col1, col2 = st.columns(2)

        with col1:
            paper1 = st.selectbox(
                "First paper",
                names,
                key="comparison_paper1"
            )

        with col2:
            paper2 = st.selectbox(
                "Second paper",
                names,
                index=1,
                key="comparison_paper2"
            )

        if st.button("Compare Papers"):
            if paper1 == paper2:
                st.warning("Please select two different papers.")
            else:
                st.info(
                    "Comparison interface ready. Backend connection is pending."
                )

    else:
        st.info(
            "Upload at least two research papers in Paper Library "
            "to prepare a comparison."
        )

# ---------------- RESEARCH INSIGHTS ----------------

elif page == "Research Insights":

    page_header(
        "Research Insights",
        "Explore topics, extracted findings and patterns across your papers."
    )

    st.markdown("""
    <div class="feature-card">
        <div class="feature-icon">🧠</div>
        <h4>Topic Classification</h4>
        <p>Identify the main research areas covered by your documents.</p>
    </div>
    """, unsafe_allow_html=True)

    st.markdown("<br>", unsafe_allow_html=True)

    st.markdown("""
    <div class="feature-card">
        <div class="feature-icon">🔑</div>
        <h4>Keyword Extraction</h4>
        <p>Discover important terms and keyphrases from research papers.</p>
    </div>
    """, unsafe_allow_html=True)

    st.markdown("<br>", unsafe_allow_html=True)

    st.markdown("""
    <div class="feature-card">
        <div class="feature-icon">📑</div>
        <h4>Research Information</h4>
        <p>Explore objectives, methodologies, datasets, results and limitations.</p>
    </div>
    """, unsafe_allow_html=True)

    st.info(
        "These sections will display real model results after backend integration."
    )

# ---------------- FOOTER ----------------

st.markdown("""
<div class="footer">
    ResearchIQ · AI Research Paper Intelligence System<br>
    Designed for smarter academic research.
</div>
""", unsafe_allow_html=True)
```
